# Lecture 20 - Object Oriented Programming

## What is an Object?

First some semantics:
 - An object is essentially a container which holds some data, and crucially some associated *methods* for working with that data. 
 - We define objects, and their behaviours, using something called a *class*. 
 - We create objects by *instantiating* classes, so, objects are *instances* of classes.

Note, these are very similar to `structures`, with associated functions attached. 

## Why do we need objects?

This is all very nice, but why bother with the overhead and confusion of objects and classes? People have been working with procedural programs for decades and they seem to work!

A few core ideas:

####  Modularity
  - Any record or action that is related to a certain entity will be all stored under a specific object
    - e.g. The number of apples and oranges in stock might be kept under an `Inventory` object
    - The actions that `Inventory` should be perform (selling apples and oranges) are also defined under the object
  - On the other hand with procedural programming, everything is stored in a single place - the main program itself it makes a program more difficult to follow once the program gets bigger


####  Separation of concerns

 - With OOP, all records and actions related to a certain entity are grouped (i.e. encapsulated) into an object
 - This makes the code easier to maintain, tests easier to write and bugs easier to track down


####  Abstraction over complex mechanisms

 - In OOP it is possible, and encouraged, to abstract (hide) away implementation details to make the interface cleaner and more easily understood

#### Inheritance

 - Often the functionality of one object has some overlap with another. Cars and busses for example are both modes of transport and may share commmon features, but need distinct behaviours.
 - Inheritance provides a way of implementing this and leads to some very powerful patterns.

### Classes 


A Class is a template to create an object

While an object refers to an actual implementation of a class with the specific information about the specific type of this object

“Class A instantiates B” means: B is an object of class A

Defining a class in Python is easy:

```python
class A:
    pass
```

Once a class is defined you can create an *instance* of that class, which is an object. In Python we do this by calling the class name as if it were a function:

```python
a_object = A() 
print(type(a_object))
```

A class can store some data (after all, an empty class isn't very interesting!):

```python
class B:

    value = 1
```

We can access variables stored in a class by writing the name of the instance followed by a dot and then the name of the variable:

```python
b_object = B()
print(b_object.value)
```

Classes can also contain functions. Functions attached to classes are called *methods*:

```python
class B(object):
    
    value = 1
    
    def _show_value(self):
        print('self.value is {}'.format(self.value))
```

The first argument to every method automatically refers to the object we're calling the method on, by convention we call that argument self.

```python
b1 = B()
b1.show_value()
b1.value = 999
b1.show_value()
```

Notice we don't have to pass the `self` argument, Python's object system does this for you.

Of course, we can pass other arguments too:

```python
class B(object):
    
    value = 1
    
    def show_value(self, another_arg):
        print('self.value is {}'.format(self.value))
        print('other arg is {}'.format(another_arg))
```

```python
b1 = B()
b1.show_value(12)
b1.value = 999
b1.show_value(132)
b2 = B()
```

Some methods are called special methods. Their names start and end with a double underscore. A particularly useful special method is `__init__`, which initializes an object.

```python
class C:
 
    def __init__(self, value):
        self.var = value
```

The `__init__` method is called when we create an instance of a class. Now when we call the class name we can pass the arguments required by `__init__`:

```python
c1 = C("Python!")

c2 = C("Hello")
print(c1.var)
print(c2.var) 
```

Methods on an object have acces to the variables defined on the object:

```python
class Counter:
    
    def __init__(self, start=0):
        self.value = start

    def increment(self):
        self.value += 1 
```

```python
counter1 = Counter()
print(counter1.value)

counter1.increment()
print(counter1.value) 
```

```python
counter2 = Counter(10)
counter2.increment()
counter2.increment()
print(counter2.value) 
```

## An example use case

Let's take a look at a slightly more useful example:

```python
class Account:

    def __init__ (self, accountHolder):
        self.balance = 0
        self.holder = accountHolder
```

And add a way of depositing money:

```python
class Account:
    def __init__ (self, accountHolder):
        self.balance = 0
        self.holder = accountHolder

    def deposit(self, amount):
        self.balance += amount
```

And withdrawing money:

```python
# Build up to this example step by step

class Account:
    def __init__ (self, accountHolder):
        self.balance = 0
        self.holder = accountHolder

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            print ("you cannot withdraw", amount)
        else:
            self.balance -= amount
```

What does this look like in practice?

```python
a = Account("Duncan")
a.deposit(100)
print(f"Balance: {a.balance}")
a.withdraw(70)
print(f"Balance: {a.balance}")
a.withdraw(40)
```

### Inheritance

 - Inheritance allows a class to inherit properties and behaviors from another class. It is a fundamental concept in OOP that promotes code reuse and establishes relationships between classes.

 - It is based on a hierarchical relationship between classes, where a derived class (also known as a subclass or child class) inherits the characteristics of a base class (also known as a superclass or parent class). 
 
 - The derived class extends the functionality of the base class by adding new features or overriding existing ones.

 - Simple inheritance is quite straight forward, but can quickly get very complex so be careful.
 - Let's take a look at creating a variation on our simple `Account` class

```python
class CreditAccount(Account):
    overdraft_rate = 0.1
    
    def __init__ (self, accountHolder, overdraft_limit=500):
        super().__init__(accountHolder)
        self.overdraft_limit = overdraft_limit
    
    def _monthly_checking():
        # Note, we're overriding the method in the base class
        if self.balance > 0.:
            self.balance += self.balance * interest_rate
        elif self.balance < 0:
            self.balance -= self.balance * overdraft_rate
    
    def withdraw(self, amount):
        # Let's adjust the logic a bit
        if (self.balance - amount) < -overdraft_limit:
            print ("you cannot withdraw", amount)
        else:
            self.balance -= amount
```
